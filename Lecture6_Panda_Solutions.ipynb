## Exercise 8: Handling Missing Values 🧼

```python
df_raw_data = pd.DataFrame({
    'Name': ['Liam', 'Noah', 'Oliver', 'Emma', 'Charlotte', 'Amelia'],
    'Age': [29, np.nan, 34, 28, np.nan, 42],
    'Department': ['Sales', 'Sales', np.nan, 'HR', np.nan, 'IT'],
    'Salary': [55000, 62000, np.nan, 58000, 72000, np.nan]
})

# SOLUTION 8
missing_counts = df_raw_data.isna().sum()
df_dropped_age = df_raw_data.dropna(subset=['Age'])

df_cleaned = df_raw_data.copy()
df_cleaned['Salary'] = df_cleaned['Salary'].fillna(df_cleaned['Salary'].median())
df_cleaned['Department'] = df_cleaned['Department'].ffill()

print("Missing counts per column:\n", missing_counts)
print("\nDataFrame with Age dropped:")
display(df_dropped_age)
print("\nFully Cleaned DataFrame:")
display(df_cleaned)
```

## Exercise 13: Summary Stats, Ranking & Window Functions 📈

## Exercise 14: GroupBy & Named Aggregations 👥📊

```python
df_sales_orders = pd.DataFrame({
    'Order_ID': range(101, 109),
    'Region': ['East', 'West', 'East', 'North', 'West', 'East', 'North', 'West'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'A'],
    'Revenue': [3500, 4200, 5100, 2100, 3800, 6200, 1800, 4900]
})

# SOLUTION 14
region_summary = df_sales_orders.groupby('Region').agg(
    total_rev=('Revenue', 'sum'),
    avg_rev=('Revenue', 'mean'),
    order_count=('Order_ID', 'count')
)

df_sales_orders['Pct_Of_Region_Rev'] = (
    df_sales_orders['Revenue'] / df_sales_orders.groupby('Region')['Revenue'].transform('sum')
) * 100

high_revenue_regions_df = df_sales_orders.groupby('Region').filter(lambda g: g['Revenue'].sum() > 10000)

print("Region Summary (Named Aggregations):")
display(region_summary)
print("\nOrders with % of Region Revenue:")
display(df_sales_orders)
print("\nHigh Revenue Regions (>10k):")
display(high_revenue_regions_df)
```

## Exercise 16: Merging, Joining & Concatenation 🔗

```python
df_cust_info = pd.DataFrame({
    'Cust_ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'City': ['NYC', 'LA', 'Chicago', 'Miami']
})

df_tx_info = pd.DataFrame({
    'Tx_ID': [901, 902, 903, 904],
    'Cust_ID': [1, 2, 2, 5],
    'Amount': [500, 300, 150, 700]
})

df_batch1 = pd.DataFrame({'Item': ['A', 'B'], 'Stock': [10, 20]})
df_batch2 = pd.DataFrame({'Item': ['C', 'D'], 'Stock': [30, 40]})

# SOLUTION 16
df_inner = pd.merge(df_cust_info, df_tx_info, on='Cust_ID', how='inner')
df_left = pd.merge(df_cust_info, df_tx_info, on='Cust_ID', how='left')
df_outer = pd.merge(df_cust_info, df_tx_info, on='Cust_ID', how='outer', indicator=True)
df_concat = pd.concat([df_batch1, df_batch2], axis=0, ignore_index=True)

print("Inner Merge:")
display(df_inner)
print("\nLeft Merge:")
display(df_left)
print("\nOuter Merge with Indicator:")
display(df_outer)
print("\nConcatenated Batches:")
display(df_concat)
```

```python
# VERIFICATION TEST 16
assert df_inner is not None and len(df_inner) == 3, f"Expected 3 rows in inner merge, got {len(df_inner)}"
assert df_left is not None and len(df_left) == 5, f"Expected 5 rows in left merge, got {len(df_left)}"
assert df_outer is not None and '_merge' in df_outer.columns, "Outer merge missing _merge indicator"
assert (df_outer['_merge'] == 'right_only').sum() == 1, "Expected 1 right_only row in outer merge"
assert df_concat is not None and len(df_concat) == 4 and list(df_concat['Item']) == ['A', 'B', 'C', 'D'], "Concatenation incorrect"

print("✅ Exercise 16 Passed!")
```

## Exercise 17: Datetime Feature Engineering & Resampling (`.dt`) ⏱️

```python
df_timestamps = pd.DataFrame({
    'Timestamp': [
        '2026-01-01 09:15:00', '2026-01-01 14:30:00', '2026-01-02 11:00:00',
        '2026-01-03 16:45:00', '2026-01-04 10:20:00', '2026-01-04 18:00:00'
    ],
    'Amount': [100, 150, 200, 350, 120, 280]
})

# SOLUTION 17
df_timestamps['Timestamp'] = pd.to_datetime(df_timestamps['Timestamp'])
df_timestamps['Year'] = df_timestamps['Timestamp'].dt.year
df_timestamps['Month'] = df_timestamps['Timestamp'].dt.month
df_timestamps['Day_Name'] = df_timestamps['Timestamp'].dt.day_name()
df_timestamps['Is_Weekend'] = df_timestamps['Timestamp'].dt.dayofweek >= 5

df_daily_resampled = df_timestamps.set_index('Timestamp').resample('D')['Amount'].sum()

display(df_timestamps)
print("\nDaily Resampled Summary:")
display(df_daily_resampled)
```

## Exercise 18: Method Chaining & Piping (`.assign()`, `.pipe()`, `.eval()`) 🔗⚡

```python
df_raw_sales = pd.DataFrame({
    'Order': ['O1', 'O2', 'O3', 'O4', 'O5'],
    'Revenue': [1000, 1500, 800, 2200, 1300],
    'Base_Cost': [600, 900, 500, 1200, 850],
    'Shipping': [50, 80, 40, 100, 60],
    'Status': ['Completed', 'Completed', 'Cancelled', 'Completed', 'Completed']
})

# SOLUTION 18
df_processed = (
    df_raw_sales
    .query("Status != 'Cancelled'")
    .assign(
        Total_Cost=lambda df: df['Base_Cost'] + df['Shipping'],
        Margin=lambda df: df['Revenue'] - (df['Base_Cost'] + df['Shipping'])
    )
    .sort_values(by='Margin', ascending=False)
)

display(df_processed)
```

```python
# VERIFICATION TEST 18
assert df_processed is not None, "df_processed is not defined"
assert len(df_processed) == 4, f"Expected 4 rows after dropping Cancelled orders, got {len(df_processed)}"
assert 'Cancelled' not in df_processed['Status'].values, "Cancelled order was not filtered out"
assert list(df_processed.columns) == ['Order', 'Revenue', 'Base_Cost', 'Shipping', 'Status', 'Total_Cost', 'Margin'] or 'Margin' in df_processed.columns, "Columns missing"
assert df_processed.iloc[0]['Order'] == 'O4' and df_processed.iloc[0]['Margin'] == 900, "Top margin order should be O4 with margin 900"
assert df_processed.iloc[-1]['Margin'] == 350, "Lowest margin should be 350"

print("✅ Exercise 18 Passed!")
```

## Exercise 19: Comprehensive Real-World E-Commerce Sales Capstone 🏆🛒

```python
np.random.seed(42)
n_orders = 300

df_ecommerce_sales = pd.DataFrame({
    'Order_ID': np.arange(1001, 1001 + n_orders),
    'Order_Date': pd.date_range('2025-01-01', periods=n_orders, freq='D'),
    'Product': np.random.choice(['Laptop', 'Smartphone', 'Tablet', 'Headphones', 'Smartwatch'], n_orders),
    'Category': np.random.choice(['Electronics', 'Accessories', 'Mobile'], n_orders),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], n_orders),
    'Customer_Segment': np.random.choice(['Consumer', 'Corporate', 'Home Office'], n_orders),
    'Units_Sold': np.random.randint(1, 10, n_orders),
    'Price_Per_Unit': np.random.choice([1200.0, 800.0, 450.0, 150.0, 250.0], n_orders),
    'Cost_Per_Unit': np.random.choice([700.0, 500.0, 280.0, 80.0, 140.0], n_orders),
    'Discount': np.random.choice([0.0, 0.05, 0.10, 0.15, 0.20], n_orders)
})

# SOLUTION 19
# 1. Feature Engineering
df_ecommerce_sales['Total_Revenue'] = (df_ecommerce_sales['Units_Sold'] * df_ecommerce_sales['Price_Per_Unit']) * (1 - df_ecommerce_sales['Discount'])
df_ecommerce_sales['Profit'] = df_ecommerce_sales['Total_Revenue'] - (df_ecommerce_sales['Units_Sold'] * df_ecommerce_sales['Cost_Per_Unit'])
df_ecommerce_sales['Order_Month'] = df_ecommerce_sales['Order_Date'].dt.month_name()

# 2. Product Performance
product_performance = df_ecommerce_sales.groupby('Product').agg(
    total_revenue=('Total_Revenue', 'sum'),
    units_sold=('Units_Sold', 'sum'),
    avg_discount=('Discount', 'mean'),
    total_profit=('Profit', 'sum')
).sort_values(by='total_revenue', ascending=False)

# 3. Regional Performance
region_performance = df_ecommerce_sales.groupby('Region').agg(
    total_revenue=('Total_Revenue', 'sum'),
    mean_profit=('Profit', 'mean')
)

# 4. Category vs Customer Segment Pivot Table
category_segment_pivot = df_ecommerce_sales.pivot_table(
    values='Total_Revenue',
    index='Category',
    columns='Customer_Segment',
    aggfunc='sum',
    margins=True
)

# 5. Price Tier Discretization
df_ecommerce_sales['Price_Tier'] = pd.cut(
    df_ecommerce_sales['Price_Per_Unit'],
    bins=[0, 300, 700, 1500],
    labels=['Budget', 'Mid-Range', 'Premium']
)

print("--- Product Performance ---")
display(product_performance)
print("\n--- Regional Performance ---")
display(region_performance)
print("\n--- Category x Customer Segment Revenue Pivot ---")
display(category_segment_pivot)
print("\n--- Price Tier Distribution ---")
print(df_ecommerce_sales['Price_Tier'].value_counts())
```
