## Exercise 5: Boolean Masking & Grade Curving 🎯

**Task:**
Given an array of student exam scores `scores = np.array([45, 88, 52, 95, 38, 71, 60, 49, 82])`:
1. Create a boolean mask `passed_mask` for scores $\ge 50$.
2. Extract all passing scores into an array `passing_scores`.
3. Count how many students passed and store the count in `num_passed`.
4. Curve the grades: Create a new array `curved_scores` where all failing scores (< 50) are replaced with `50`, while passing scores remain unchanged.

> **Hint:** Use boolean comparison operators and `np.where(condition, x, y)`.


````python
scores = np.array([45, 88, 52, 95, 38, 71, 60, 49, 82])

# YOUR CODE HERE
passed_mask = scores>=50
passing_scores = scores[scores>=50]
num_passed = sum(passed_mask)
curved_scores = np.where(scores<50, 50, scores)
print(passed_mask)
print("Passing scores:", passing_scores)
print("Number of students passed:", num_passed)
print("Curved scores:", curved_scores)
````

````python
# VERIFICATION TEST 5
np.testing.assert_array_equal(passing_scores, np.array([88, 52, 95, 71, 60, 82]))
assert num_passed == 6, f"Expected 6, got {num_passed}"
np.testing.assert_array_equal(curved_scores, np.array([50, 88, 52, 95, 50, 71, 60, 50, 82]))

print("✅ Exercise 5 Passed!")
````

## Exercise 6: Reshaping and Stacking Arrays 🧩

**Task:**
1. Start with a 1D array `arr_12` containing numbers from `0` to `11`.
2. Reshape `arr_12` into a `3x4` matrix named `mat_3x4`.
3. Create a `3x2` matrix of ones named `ones_3x2`.
4. Horizontally stack `mat_3x4` and `ones_3x2` into a `3x6` matrix named `stacked_mat`.
5. Flatten `stacked_mat` into a 1D array named `flat_arr`.

> **Hint:** Use `.reshape()`, `np.hstack()`, and `.flatten()`.

## Exercise 7: Axis Statistics 📊

**Task:**
Below is a matrix of exam scores for `4` students across `3` subjects:
```python
exam_data = np.array([
    [85, 90, 78],  # Student 0
    [60, 75, 80],  # Student 1
    [92, 88, 95],  # Student 2
    [70, 65, 72]   # Student 3
])
```
1. Calculate the mean score for **each student** (across subjects) and store in `student_means` (shape: `(4,)`).
2. Calculate the highest score for **each subject** (across students) and store in `subject_maxes` (shape: `(3,)`).
3. Find the overall maximum score in the entire matrix `overall_max` and the index of the highest scoring student (row index) `best_student_idx`.

> **Hint:** Remember `axis=1` operates across columns (for each row), and `axis=0` operates across rows (for each column).


````python
exam_data = np.array([
    [85, 90, 78],
    [60, 75, 80],
    [92, 88, 95],
    [70, 65, 72]
])

# YOUR CODE HERE
student_means = np.mean(exam_data, axis=1)
subject_maxes = np.max(exam_data, axis=0)
overall_max = np.max(exam_data)
best_student_idx = np.argmax(student_means)

print("Student Means:", student_means)
print("Subject Maxes:", subject_maxes)
print("Overall Max:", overall_max)
print("Best Student Index:", best_student_idx)
````

## Exercise 8: Data Standardization via Broadcasting 📡

**Task:**
Standardization (Z-score normalization) transforms dataset features to have `mean = 0` and `std = 1` using the formula:
$$Z = \frac{X - \mu}{\sigma}$$
where $\mu$ is the mean of each column and $\sigma$ is the standard deviation of each column.

Given dataset `X` of shape `(4, 2)`:
1. Compute `col_means` (mean of each column).
2. Compute `col_stds` (std deviation of each column).
3. Compute standardized data `X_std` using NumPy broadcasting.

> **Hint:** NumPy automatically broadcasts shape `(2,)` vectors across `(4, 2)` matrices!


````python
X = np.array([
    [10.0, 200.0],
    [20.0, 400.0],
    [30.0, 100.0],
    [40.0, 300.0]
])

# YOUR CODE HERE
col_means = np.mean(X, axis=0)
col_stds = np.std(X, axis=0)
X_std = (X-col_means)/col_stds

print("Column Means:", col_means)
print("Column Stds:", col_stds)
print("Standardized X:\n", X_std)
````

````python
# VERIFICATION TEST 8
np.testing.assert_allclose(col_means, np.array([25.0, 250.0]))
np.testing.assert_allclose(col_stds, np.array([11.18033989, 111.80339887]))

# Standardized data should have mean ~ 0 and std ~ 1
np.testing.assert_allclose(np.mean(X_std, axis=0), np.array([0.0, 0.0]), atol=1e-7)
np.testing.assert_allclose(np.std(X_std, axis=0), np.array([1.0, 1.0]), atol=1e-7)

print("✅ Exercise 8 Passed!")
````

## Exercise 10: Cleaning Missing Data (NaN) 🧹

**Task:**
You are given an array with missing entries: `data_nan = np.array([12.0, np.nan, 18.0, 24.0, np.nan, 30.0])`.
1. Calculate the mean of valid (non-NaN) entries and store in `valid_mean`.
2. Create a clean array `data_imputed` where all `NaN` values are replaced by `valid_mean`.

> **Hint:** Use `np.nanmean()` and `np.where(np.isnan(...), ...)` or `np.nan_to_num()`.


````python
data_nan = np.array([12.0, np.nan, 18.0, 24.0, np.nan, 30.0])

# YOUR CODE HERE
valid_mean = np.nanmean(data_nan)
data_imputed = data_nan[~np.isnan(data_nan)]

print("Valid Mean:", valid_mean)
print("Imputed Data:", data_imputed)
````

````python
# VERIFICATION TEST 10
assert valid_mean == 21.0, f"Expected mean 21.0, got {valid_mean}"
assert not np.isnan(data_imputed).any(), "data_imputed still contains NaN values!"
np.testing.assert_allclose(data_imputed, np.array([12.0, 21.0, 18.0, 24.0, 21.0, 30.0]))

print("✅ Exercise 10 Passed!")
````

## Exercise 12: Real-World Challenge - Synthetic Image Filter 🖼️

**Background:**
In digital image processing, a grayscale image is represented as a 2D NumPy array of pixel values ranging from `0` (black) to `255` (white).

**Task:**
You are given an 8x8 synthetic image:
```python
np.random.seed(100)
img = np.random.randint(0, 256, (8, 8), dtype=np.uint8)
```
Complete the following image processing tasks:
1. **Brightening:** Create `bright_img` by scaling pixel intensities by `1.4`. Clip values so they stay within range `[0, 255]`, and cast to `np.uint8`.
2. **Binary Thresholding:** Create `binary_img` where pixels $> 128$ become `255`, and pixels $\le 128$ become `0` (dtype `np.uint8`).
3. **Horizontal Flip:** Flip `binary_img` horizontally to produce `flipped_img`.

> **Hint:** Use `np.clip()`, `np.where()`, and `np.fliplr()`.


````python
np.random.seed(100)
img = np.random.randint(0, 256, (8, 8), dtype=np.uint8)

# YOUR CODE HERE
bright_img = np.clip(img*1.4, 0,255).astype(np.uint8)
binary_img = np.where(img>128, 255, 0)
flipped_img = np.fliplr(binary_img)

print("Original Image:\n", img)
print("Brightened Image:\n", bright_img)
print("Binary Image:\n", binary_img)
print("Flipped Binary Image:\n", flipped_img)
````

````python
# VERIFICATION TEST 12
np.random.seed(100)
expected_img = np.random.randint(0, 256, (8, 8), dtype=np.uint8)

expected_bright = np.clip(expected_img * 1.4, 0, 255).astype(np.uint8)
expected_binary = np.where(expected_img > 128, 255, 0).astype(np.uint8)
expected_flipped = np.fliplr(expected_binary)

np.testing.assert_array_equal(bright_img, expected_bright)
np.testing.assert_array_equal(binary_img, expected_binary)
np.testing.assert_array_equal(flipped_img, expected_flipped)

print("🎉 Congratulations! All 12 Exercises Passed Successfully!")
````
